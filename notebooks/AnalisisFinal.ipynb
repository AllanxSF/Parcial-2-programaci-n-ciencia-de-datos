# Análisis Final - Modelo K-Means

## 1. Impacto del escalado

Se entrenaron dos modelos: uno utilizando los datos originales y otro utilizando datos escalados mediante StandardScaler.

La comparación entre ambos modelos mostró cambios en la asignación de registros a distintos grupos, lo que indica que el proceso de normalización tuvo un efecto importante en el agrupamiento.

Esto ocurre debido a que K-Means utiliza distancias entre observaciones y variables con escalas mayores pueden influir más en la formación de clusters.

## 2. Calidad del agrupamiento

La calidad de los grupos fue evaluada utilizando el coeficiente Silhouette.

El valor obtenido fue aproximadamente:

Silhouette Score = 0.094

Este valor se encuentra cercano a 0, indicando una separación débil entre los grupos encontrados.

Los resultados sugieren que los registros presentan características relativamente similares y que no existen agrupaciones naturales claramente definidas dentro del conjunto de datos.

## 3. Optimización del modelo

Se evaluaron distintos valores de K utilizando el método del codo y el coeficiente Silhouette.

Los resultados obtenidos fueron:

- K=2 → 0.094
- K=3 → 0.088
- K=4 → 0.085
- K=5 → 0.079
- K=6 → 0.077
- K=7 → 0.078

El mejor desempeño fue obtenido con K=2, aunque los valores continuaron siendo bajos, indicando una estructura de agrupamiento débil.

## 4. Relación con la variable objetivo

Se realizó una comparación entre los clusters generados y la variable target_calidad.

Los resultados mostraron una distribución relativamente uniforme entre las categorías:

- Desecho
- Exportación
- Local

Esto indica que los grupos generados por K-Means no representan de forma clara la clasificación real de calidad del producto.

## Conclusión general

Los resultados mostraron que el proceso de escalado influyó en la asignación de registros a los clusters, evidenciando la importancia de normalizar los datos antes del entrenamiento.

La optimización realizada mediante el coeficiente Silhouette determinó que el mejor valor fue K=2; sin embargo, el puntaje obtenido fue bajo (≈0.094), indicando una separación débil entre los grupos encontrados.

Además, la comparación entre los clusters y la variable `target_calidad` mostró que las categorías `Desecho`, `Exportación` y `Local` se distribuyen de forma similar entre los grupos, por lo que K-Means no logró representar claramente la clasificación real de los datos.

Aunque el modelo presenta limitaciones frente a modelos supervisados para tareas de clasificación, resultó útil como herramienta de análisis exploratorio, permitiendo identificar patrones generales y comprender mejor el comportamiento del conjunto de datos.
