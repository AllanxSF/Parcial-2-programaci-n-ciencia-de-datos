# Análisis Final - Modelo K-Means

## 1. Impacto del escalado

Se entrenaron dos modelos: uno utilizando los datos originales y otro utilizando datos escalados mediante StandardScaler.

La comparación entre ambos modelos mostró cambios en la asignación de registros a distintos grupos, lo que indica que el proceso de normalización tuvo un efecto importante en el agrupamiento.

Esto ocurre debido a que K-Means utiliza distancias entre observaciones y variables con escalas mayores pueden influir más en la formación de clusters.

## 2. Calidad del agrupamiento

La calidad de los grupos fue evaluada utilizando el coeficiente Silhouette.

El valor obtenido fue aproximadamente:

Silhouette Score = 0.094

Este valor se encuentra cercano a 0, indicando una separación débil entre los grupos encontrados.

Los resultados sugieren que los registros presentan características relativamente similares y que no existen agrupaciones naturales claramente definidas dentro del conjunto de datos.

## 3. Optimización del modelo

Se evaluaron distintos valores de K utilizando el método del codo y el coeficiente Silhouette.

Los resultados obtenidos fueron:

- K=2 → 0.094
- K=3 → 0.088
- K=4 → 0.085
- K=5 → 0.079
- K=6 → 0.077
- K=7 → 0.078

El mejor desempeño fue obtenido con K=2, aunque los valores continuaron siendo bajos, indicando una estructura de agrupamiento débil.

## 4. Relación con la variable objetivo

Se realizó una comparación entre los clusters generados y la variable target_calidad.

Los resultados mostraron una distribución relativamente uniforme entre las categorías:

- Desecho
- Exportación
- Local

Esto indica que los grupos generados por K-Means no representan de forma clara la clasificación real de calidad del producto.

## Conclusión general

Los resultados mostraron que el proceso de escalado influyó en la asignación de registros a los clusters, evidenciando la importancia de normalizar los datos antes del entrenamiento.

La optimización realizada mediante el coeficiente Silhouette determinó que el mejor valor fue K=2; sin embargo, el puntaje obtenido fue bajo (≈0.094), indicando una separación débil entre los grupos encontrados.

Además, la comparación entre los clusters y la variable `target_calidad` mostró que las categorías `Desecho`, `Exportación` y `Local` se distribuyen de forma similar entre los grupos, por lo que K-Means no logró representar claramente la clasificación real de los datos.

Aunque el modelo presenta limitaciones frente a modelos supervisados para tareas de clasificación, resultó útil como herramienta de análisis exploratorio, permitiendo identificar patrones generales y comprender mejor el comportamiento del conjunto de datos.

# Análisis Final - Modelo Decision-Tree

## Conclusiones - Decision Tree

- **Accuracy:** 36.67% — desempeño global bajo; el árbol acierta poco más que un clasificador aleatorio en 3 clases.
- **Patrón en la matriz de confusión:** fuerte sesgo a predecir `Local`; `Desecho` no es detectada y `Exportación` se confunde con `Local`.
- **Optimización:** GridSearchCV y RandomizedSearchCV mostraron diferencias marginales en CV y test; ninguna mejoría sustancial en la generalización.
- **Interpretación del árbol:** las primeras divisiones (`remainder_nitrogen`, `remainder_humedad_suelo`, `remainder_radiacion_solar`) son las más determinantes; limitar la profundidad a 3 mantiene interpretabilidad pero muestra fronteras de clase superpuestas.

En conjunto: el modelo identifica señales útiles pero requiere acciones (balanceo de clases, ingeniería de características o modelos ensembles) para mejorar la detección de `Desecho` y `Exportación`.

### Qué podemos mejorar

- Balanceo de clases: usar `class_weight='balanced'`, SMOTE o undersampling dentro del pipeline (nunca antes de CV).
  - Por qué: el modelo está sesgado hacia la clase mayoritaria (`Local`); balancear las clases permite que el clasificador vea suficientes ejemplos de `Desecho` para aprender sus patrones.

- Cambiar la métrica de optimización a `f1_macro` y usar `StratifiedKFold`.
  - Por qué: `accuracy` puede ocultar mal rendimiento en clases minoritarias; `f1_macro` y validación estratificada aseguran que el proceso de búsqueda priorice desempeño por clase.

- Probar modelos ensemble (RandomForest, GradientBoosting/XGBoost) con `class_weight` o ajuste de pesos.
  - Por qué: los ensemble suelen capturar relaciones no lineales y reducir varianza, lo que ayuda cuando un árbol simple no separa bien las clases.

- Ingeniería de features: crear interacciones, agregados o transformaciones que separen mejor `Desecho` vs `Exportación`.
  - Por qué: puede que no existan variables individuales suficientemente discriminantes; combinarlas o transformarlas puede mejorar la señal para el clasificador.

- Calibración y ajuste de umbral en probabilidades para priorizar recall en `Desecho` si es crítico.
  - Por qué: calibrar probabilidades permite ajustar el trade-off entre precisión y recall; bajar el umbral para `Desecho` puede aumentar su detección con costo en más falsos positivos.

- Revisar la calidad de las etiquetas y, si es posible, recolectar más ejemplos de clases minoritarias.
  - Por qué: errores de etiquetado o pocos ejemplos dificultan el aprendizaje; más y mejores datos suelen ser la solución más efectiva a largo plazo.
